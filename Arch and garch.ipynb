```python
from arch import arch_model
import pandas as pd

def auto_arch(data, max_p=10, criterion="AIC"):

    data = data.dropna() * 100

    results = []

    for p in range(1, max_p + 1):

        model = arch_model(
            data,
            mean="Constant",
            vol="ARCH",
            p=p,
            dist="normal"
        )

        result = model.fit(disp="off")

        results.append({
            "p": p,
            "AIC": result.aic,
            "BIC": result.bic,
            "LogLik": result.loglikelihood,
            "Model": result
        })

    results_df = pd.DataFrame(results)

    # Find the row with the lowest AIC/BIC
    best_index = results_df[criterion].idxmin()

    best_result = results_df.loc[best_index]

    return results_df, best_result
```

```python
nifty_models, nifty_best = auto_arch(
    merged_df["Nifty_LogReturn"],
    max_p=10,
    criterion="AIC"
)

print(nifty_models[["p", "AIC", "BIC", "LogLik" ]])

print("\nBest ARCH model:")
print(nifty_best[["p", "AIC", "BIC", "LogLik"]])
```

```python
from arch import arch_model

data = merged_df["Nifty_LogReturn"].dropna() * 100

# -----------------------------
# 1. Best ARCH(p), p = 1 to 10
# -----------------------------

best_arch_aic = float("inf")
best_arch_p = None
best_arch_model = None

for p in range(1, 11):

    model = arch_model(
        data,
        mean="Constant",
        vol="ARCH",
        p=p,
        dist="normal"
    )

    result = model.fit(disp="off")

    if result.aic < best_arch_aic:
        best_arch_aic = result.aic
        best_arch_p = p
        best_arch_model = result


# -----------------------------
# 2. Best GARCH(p,q)
# -----------------------------

best_garch_aic = float("inf")
best_garch_p = None
best_garch_q = None
best_garch_model = None

for p in range(1, 3):
    for q in range(1, 3):

        model = arch_model(
            data,
            mean="Constant",
            vol="GARCH",
            p=p,
            q=q,
            dist="normal"
        )

        result = model.fit(disp="off")

        if result.aic < best_garch_aic:
            best_garch_aic = result.aic
            best_garch_p = p
            best_garch_q = q
            best_garch_model = result


# -----------------------------
# 3. Compare
# -----------------------------

print("=" * 60)
print("NIFTY VOLATILITY MODEL COMPARISON")
print("=" * 60)

print(
    f"Best ARCH: ARCH({best_arch_p})"
)
print(
    f"ARCH AIC: {best_arch_aic:.3f}"
)

print()

print(
    f"Best GARCH: GARCH({best_garch_p},{best_garch_q})"
)
print(
    f"GARCH AIC: {best_garch_aic:.3f}"
)

print()

if best_arch_aic < best_garch_aic:
    print("FINAL CHOICE: ARCH")
else:
    print("FINAL CHOICE: GARCH")
```
