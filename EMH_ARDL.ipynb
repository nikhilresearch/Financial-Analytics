```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Download 5 years of daily data

reliance_ts = yf.download(
    "RELIANCE.NS",
    start="2020-01-01",
    end="2026-09-09",
    progress=False
)

canara_ts = yf.download(
    "CANBK.NS",
    start="2020-01-01",
    end="2026-09-09",
    progress=False
)

nifty_ts = yf.download(
    "^NSEI",
    start="2020-01-01",
    end="2026-09-09",
    progress=False
)
```

```python
nifty_ts.head()
```

```python
print(nifty_ts.columns)
```

```python
reliance_ts = reliance_ts[['Close']]
canara_ts = canara_ts[['Close']]
nifty_ts = nifty_ts[['Close']]

```

```python
reliance_ts.columns = ['Reliance']
canara_ts.columns = ['Canara']
nifty_ts.columns = ['Nifty']
```

```python
nifty_ts.head()
```

```python
merged_df = pd.concat(
    [reliance_ts, canara_ts, nifty_ts],
    axis=1
)

merged_df.head()
```

```python
print(merged_df.columns)
```

```python
import numpy as np

log_returns = np.log(
    merged_df[['Reliance', 'Canara', 'Nifty']] /
    merged_df[['Reliance', 'Canara', 'Nifty']].shift(1)
).dropna()

log_returns.head()
```

```python
import statsmodels.api as sm
```

```python
X = sm.add_constant(log_returns["Nifty"])
y = log_returns["Reliance"]
```

```python
model_reliance = sm.OLS(y, X).fit()
```

```python
print(model_reliance.summary())
```

```python
X_c = sm.add_constant(log_returns["Nifty"])
y_c = log_returns["Canara"]
```

```python
model_canara = sm.OLS(y_c, X_c).fit()
```

```python
print(model_canara.summary())
```

```python
canara = yf.download(
    "CANBK.NS",
    start="2020-01-01",
    end="2026-09-01",
    interval="1mo",
    auto_adjust=False
)

nifty = yf.download(
    "^NSEI",
    start="2020-01-01",
    end="2026-09-01",
    interval="1mo",
    auto_adjust=False
)
```

```python
nifty.head()
```

```python
canara = canara[['Close']]
nifty = nifty[['Close']]
```

```python
canara.columns=['canara']
nifty.columns=['nifty']
nifty.head()
```

```python
monthly_df = pd.concat(
    [ canara, nifty],
    axis=1
)

monthly_df.head()
```

```python
monthly_df['Rf'] = 0.5
```

```python
monthly_df.head()
```

```python
monthly_df = np.log(
    monthly_df[['canara', 'nifty']] /
    monthly_df[[ 'canara', 'nifty']].shift(1)
).dropna()

monthly_df.head()
```

```python
monthly_df['Rf'] = 0.5
```

```python
monthly_df.head()
```

```python
monthly_df['Canara_Excess'] = monthly_df['canara'] - monthly_df['Rf']
monthly_df['Nifty_Excess'] = monthly_df['nifty'] - monthly_df['Rf']
```

```python
monthly_df.head()
```

```python
X_m = sm.add_constant(monthly_df["nifty"])
y_m = monthly_df["canara"]
```

```python
monthly_canara = sm.OLS(y_m, X_m).fit()
```

```python
print(monthly_canara.summary())
```

ARDL

```python
from statsmodels.tsa.ardl import ARDL
```

```python
ardl_data = merged_df[['Canara', 'Nifty']].dropna()
```

```python
from statsmodels.tsa.ardl import ARDL

model_ardl = ARDL(
    ardl_data['Canara'],
    lags=1,
    exog=ardl_data[['Nifty']],
    order=1
)

result_ardl = model_ardl.fit()

print(result_ardl.summary())
```

```python
models = {}

for p in range(1, 4):
    for q in range(0, 4):
        
        model = ARDL(
            ardl_data['Canara'],
            lags=p,
            exog=ardl_data[['Nifty']],
            order=q
        )
        
        result = model.fit()
        
        models[f'ARDL({p},{q})'] = result
```

```python
aic_table = pd.DataFrame({
    name: {
        'AIC': result.aic,
        'BIC': result.bic
    }
    for name, result in models.items()
}).T

print(aic_table.sort_values('AIC'))
```

```python
model_ardl2 = ARDL(
    ardl_data['Canara'],
    lags=3,
    exog=ardl_data[['Nifty']],
    order=2
)

result_ardl2 = model_ardl2.fit()

print(result_ardl2.summary())
```

```python
model_ardl3 = ARDL(
    ardl_data['Canara'],
    lags=1,
    exog=ardl_data[['Nifty']],
    order=1
)

result_ardl3 = model_ardl3.fit()

print(result_ardl3.summary())
```
