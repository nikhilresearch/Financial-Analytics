```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Download 5 years of daily data

reliance_ts = yf.download(
    "RELIANCE.NS",
    start="2020-01-01",
    end="2026-08-12",
    progress=False
)

canara_ts = yf.download(
    "CANBK.NS",
    start="2020-01-01",
    end="2026-08-12",
    progress=False
)
```

```python
reliance_ts.tail()
```

```python
reliance_ts = reliance_ts[['Close']]
canara_ts = canara_ts[['Close']]

reliance_ts.columns = ['Reliance']
canara_ts.columns = ['Canara']
```

```python
reliance_ts.head()
```

```python
#merge both the dataframes
merged_df = pd.merge(
    reliance_ts,
    canara_ts,
    left_index=True,
    right_index=True,
    how="inner"
)
```

```python
merged_df.head()
```

```python
#log return of price
merged_df["Reliance_LogReturn"] = np.log(
    merged_df["Reliance"] /
    merged_df["Reliance"].shift(1)
)

merged_df["Canara_LogReturn"] = np.log(
    merged_df["Canara"] /
    merged_df["Canara"].shift(1)
)
```

```python
#hypothesis H0= there is no ARCH exist
from statsmodels.stats.diagnostic import het_arch

for col in return_cols:

    data = merged_df[col].dropna() * 100

    test = het_arch(
        data,
        nlags=10
    )

    print("\n", "=" * 50)
    print(col)
    print("=" * 50)

    print("LM Statistic :", test[0])
    print("LM p-value   :", test[1])
    print("F Statistic  :", test[2])
    print("F p-value    :", test[3])
```

```python
from arch import arch_model

arch_results = {}

for col in [
    "Nifty_LogReturn",
    "Reliance_LogReturn",
    "Canara_LogReturn"
]:

    data = merged_df[col].dropna() * 100

    model = arch_model(
        data,
        mean="Constant",
        vol="ARCH",
        p=1,
        dist="normal"
    )

    result = model.fit(disp="off")

    arch_results[col] = result

    print("\n", "="*70)
    print(col)
    print("="*70)

    print(result.summary())
```
